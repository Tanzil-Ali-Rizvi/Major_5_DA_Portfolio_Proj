```python
import pandas as pd
import statsmodels.api as sm

# 1. Loading Data
data = pd.read_csv('data/logistics_delivery_data.csv')

# 2. Preparing Regression (One-hot encoding for Hubs)
X = pd.get_dummies(data[['distance_miles', 'hub']], drop_first=True).astype(float)
X = sm.add_constant(X)
y = data['delay_hours']

# 3. Fit Model
model = sm.OLS(y, X).fit()

# 4. Saving Summary as Text for README
with open("visuals/regression_summary.txt", "w") as f:
    f.write(model.summary().as_text())

print(model.summary())
```
